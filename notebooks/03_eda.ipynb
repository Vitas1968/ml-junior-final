```
import pandas as pd
from pathlib import Path

PROJECT_ROOT = Path("..").resolve()
data_path = PROJECT_ROOT / "data" / "processed" / "dataset.pkl"

X = pd.read_pickle(data_path)
X.shape
```

```


PROJECT_ROOT = Path("..").resolve()
target_path = PROJECT_ROOT / "data" / "target" / "train_target.csv"

y = pd.read_csv(target_path)
y.shape
```

```
y.head()
```

```
y["id"].nunique(), y.shape[0]
```

```
df_full = X.merge(y, on="id", how="inner")
df_full.shape
```

```
# проверка, что строки не потерялись
assert df_full.shape[0] == X.shape[0], "Потеря строк при merge"
assert "flag" in df_full.columns, "Таргет не добавился"
# проверка наличия таргета
df_full.columns
```
